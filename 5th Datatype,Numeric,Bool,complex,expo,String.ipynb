# Data Types

Numeric

```python
v1 = 10 # Integer data type
print(v1)
print(type(v1))   # type of object
print(v1, "is Ineger?", isinstance(v1,int))  # v1 is an instance of int
```

```python
v2 = 72.66  # Flot data type
print(v2)
print(type(v2))   #type of object
print(v2,"is float?",isinstance(v2,float))  #v2 is an instance of float
```

```python
v3 = 20+5j   #Complex data type
print(v3)
print(v3," is complex?",isinstance(v3,complex))  #v3 is an instance of complex
```

```python
import sys 
sys.getsizeof(int())   #size of integer object in bytes
```

```python
sys.getsizeof(float())   #size of float object in bytes
```

```python
sys.getsizeof(complex())   #size of complex object in bytes
```
