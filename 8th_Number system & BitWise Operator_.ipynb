# 1. Binary number system (bin) 

only Value 1 & 2 

```python
# Decimal to Binary conversion 

print ('bin(55)=', bin(55)) 


# Binary To decimal Converstion 

print ('int(0b110111)=',int(0b110111))


#bool to binary Conversion 

print('bin(True)=',bin(True)) 
print('bin(False)=',bin(False))0
```

# 2. Octal Number System (oct) 

```python
#Decimal to Octal number System conversion 

print ('oct(45)=', oct(45))
print ('int(0o55)=', int(0o55))
```

```python
#bool to Octal Number Conversion 

print('oct(True)=',oct(True))
print('oct(False)=',oct(False))
```

# 3. Decimal number system (int)

Any number you type without a prefix is automatically treated as a decimal (base-10) value


```python
print('print(25)=', 25)
```

# 4. Hexadecimal(hex)

```python
print('hex(95)=',hex(95)) 
print ('int(0x5f)=',int(0x5f))
```

```python
#bool to binary Conversion 

print('hex(True)=',hex(True))
print('hex(False)=',hex(False))
```

# Bit wise operator
          1. Complement (~) 
		  2. Bitwise and (&)
		  3. Bitwise or (|)
		  4. xor (xclusive or) ^ 
		  5. left shift << 
          6. right shift >> 

# 1. Complement (~)

```python
print('~ 120 = ', ~120) 
print('~(-120)= ', ~(-120))
```

#  2. Bitwise and (&)

Rule: output bit = 1 only if both input bits = 1

```python
print ('12 (0b1100) & 13 (0b1101) =', 12 & 13)
```

# 3. Bitwise or (|)

```python
print(bin(25))
print(bin(65))


print ('25 | 65 = ', 25 | 65)

print(bin (89))

print('True | False=', True | False)
print('False | False=', False | False)
print('True | True =', True | True)
```

#  4. xor (xclusive or) ^



```python
print( bin(75))
print(bin(45))
 
Result = 75 ^ 45

print (' 75  ^ 45 =', Result)

print(bin(Result))
```

# 5. left shift ( << )
( we are gaining bit <<2 ( 2bit (2) if 1bit (1 zero i will gain) 

```python
n = 3 #0b11
result = n << 2
print('n<<2 =', result)

print(bin(result))
```

# 6. right shift (>>)
( we will loos the bit ) >>1 ( 1bit we will loose) >>2 -- 2bit we will loose 

```python
n = 97 #0b1100001
result = n >> 2

print('n >> 2 =', result)

print(bin(result))
```
