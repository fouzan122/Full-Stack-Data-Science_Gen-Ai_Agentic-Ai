 # syntax: string[start : end : step]

 start: the index to start slicing, default value is 0.
 end: the index to stop ,default value is length of string.
 step: How much to increment the index after each character.

```python
my_name = "Fouzan"      # index: 0123
```

```python
print(my_name[0:3])   # step:1
```

```python
print(my_name[0:4:1])  # step:2
```

```python
print(my_name[0:4:2])   # step:3
```

```python
print(my_name[0:2])     # first 2 chars
```
